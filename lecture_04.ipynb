```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

N = 1000
x = np.linspace(-np.pi,np.pi,N)

fig=plt.figure(figsize=(12,8), dpi= 60, facecolor='w', edgecolor='k')
plt.plot(x, x**2, label = "$y(x)=x^2$")
plt.xlabel("x")
plt.ylabel("y")
plt.ylim((-1,10))
plt.legend()
plt.show()
```
